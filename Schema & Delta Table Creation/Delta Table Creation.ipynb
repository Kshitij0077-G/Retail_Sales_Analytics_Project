```python
# Configuration list defining our raw files and target Delta tables
INGESTION_CONFIG = [
    {
        "file_name": "dim_customers.csv",
        "table": "dim_customers"
    },
    {
        "file_name": "dim_products.csv",
        "table": "dim_products"
    },
    {
        "file_name": "fact_sales.csv",
        "table": "dim_fact_sales"
    }
]
```

```python
# Base Unity Catalog Volume path
volume_path = "/Volumes/retail_sales_analytics/analytics/raw_files"

# Loop through each configuration item to ingest and write as Delta tables
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['file_name']} → retail_sales_analytics.analytics.{item['table']}")

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(f"{volume_path}/{item['file_name']}")
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"retail_sales_analytics.analytics.{item['table']}")
    )

print("\nAll tables successfully ingested using configuration-driven ingestion!")
```
