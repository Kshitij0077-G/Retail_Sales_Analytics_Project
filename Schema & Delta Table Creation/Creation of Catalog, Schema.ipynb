```python
# 1. Create the dedicated Unity Catalog and Schema
spark.sql("CREATE CATALOG IF NOT EXISTS retail_sales_analytics;")
spark.sql("CREATE SCHEMA IF NOT EXISTS retail_sales_analytics.analytics;")

# 2. Set the active catalog and schema context
spark.sql("USE CATALOG retail_sales_analytics;")
spark.sql("USE SCHEMA analytics;")

# 3. Create a managed Unity Catalog Volume for raw file storage
spark.sql("CREATE VOLUME IF NOT EXISTS retail_sales_analytics.analytics.raw_files;")

print("Retail Sales Analytics environment and raw files volume successfully initialized!")
```
